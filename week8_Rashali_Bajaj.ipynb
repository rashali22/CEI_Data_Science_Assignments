# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
import json

# 🤖 AGENT FUNCTION

def agent(query: str):
    # Validate input
    if not isinstance(query, str):
        return {
            "type": "error",
            "result": "Invalid input: query must be a string."
        }

    query_lower = query.lower().strip()

    # Handle empty input
    if not query_lower:
        return {
            "type": "error",
            "result": "Invalid input: query cannot be empty."
        }

    try:
        # Route calculation queries
        if "calculate" in query_lower:
            expression = query_lower.split("calculate", 1)[1].strip()

            if not expression:
                return {
                    "type": "error",
                    "result": "Invalid calculation: mathematical expression is missing."
                }

            result = calculator(expression)

            # Calculator already reports calculation failure
            if result == "Error in calculation":
                return {
                    "type": "error",
                    "result": "Invalid mathematical expression."
                }

            return {
                "type": "calculation",
                "result": result
            }

        # Route keyword extraction queries
        elif "keywords" in query_lower:
            text = query_lower.split("keywords", 1)[1].strip()

            if not text:
                return {
                    "type": "error",
                    "result": "Invalid keyword request: text is missing."
                }

            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }

        # Fallback for general queries
        else:
            return {
                "type": "general",
                "result": "This is a general query."
            }

    except Exception as e:
        return {
            "type": "error",
            "result": f"Error processing query: {str(e)}"
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print(json.dumps(agent(q), indent=2))
    print("-" * 50)
````

### Additional Validation Tests

````python
test_queries = [
    "Calculate 100 / 4",
    "Calculate hello",
    "Calculate",
    "keywords",
    "Hello, how are you?"
]

for q in test_queries:
    print("Query:", q)
    print(json.dumps(agent(q), indent=2))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")

    if user_input.lower() == "exit":
        break

    print("Response:")
    print(json.dumps(agent(user_input), indent=2))
````
